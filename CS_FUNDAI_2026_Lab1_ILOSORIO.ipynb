## Simple Intelligent Agent

This simple agent receives a room condition as a percept and produces an action.

The agent follows a condition-action rule:

- if the room is dirty, the agent cleans.
- if the room is clean, the angent rests.
- if the room condition is unknown, the agent inspects.

```python
def simple_reflex_agent(room_condition):
  """
  A simple reflex agent that decides an action based on the room condition.
  """
  room_condition = room_condition.strip(). lower()

  if room_condition == "dirty":
    return "CLEAR"
  elif room_condition == "clean":
    return "REST"
  else:
    return "INSPECT"
```

```python
def simple_reflex_agent(room_condition):
  """
  A simple reflex agent that decides an action based on the room condition.
  """
  room_condition = room_condition.strip().lower()

  if room_condition == "dirty":
    return "CLEAR"
  elif room_condition == "clean":
    return "REST"
  else:
    return "INSPECT"

percepts = ["dirty" , "clean" , "dirty" , "unknown"]

print("Simple Reflex Agent Test")
print("-" * 30)

for i, percept in enumerate(percepts, start=1):
  action = simple_reflex_agent(percept)
  print(f"Percept {i}: {percept:<8} -> Action: {action}")
```

## Performace Mesure

A simple performance measure gives a score when the agent chooses the correct action for the given percept

```python
score = 0
total = len(percepts)

for percept in percepts:
  action = simple_reflex_agent(percept)

  if percept == "dirty" and action == "CLEAR":
    score += 1
  elif percept == "clean" and action == "REST":
    score += 1
  elif percept == "unknown" and action == "INSPECT":
    score += 1

print(f"Agent score: {score}/{total}")
```
